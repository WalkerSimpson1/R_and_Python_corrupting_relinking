```python
judges_merge = pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/judge_case_merge_key.csv")
judges_clean = pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/judges_clean.csv")

judges_clean = judges_clean.with_columns(pl.col(["start_date", "end_date"]).str.to_date(format="%Y-%m-%d", strict=False))

filing_judges_clean = judges_clean.select(pl.col(['ddl_judge_id', 'female_judge', 'start_date', 'end_date']).name.suffix("_filing"))
decision_judges_clean = judges_clean.select(pl.col(['ddl_judge_id', 'female_judge', 'start_date', 'end_date']).name.suffix("_decision"))
```

```python
cases_labeled = cases_labeled.join(judges_merge, on="ddl_case_id", how="left")

cases_labeled = cases_labeled.join(filing_judges_clean, left_on="ddl_filing_judge_id", right_on="ddl_judge_id_filing", how="left")
cases_labeled = cases_labeled.join(decision_judges_clean, left_on="ddl_decision_judge_id", right_on="ddl_judge_id_decision", how="left")
```

```python
pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/acts_sections.csv").select(pl.col("section").value_counts(sort=True)).unnest("section").collect()
```

```python
act_key = pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/act_key.csv")
section_key = pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/section_key.csv")
acts_sections = pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/acts_sections.csv")
```

```python
act_key = act_key.select(pl.col(["act", "act_s"]))
section_key = section_key.select(pl.col(["section", "section_s"]))
```

```python
acts_sections = acts_sections.with_columns(pl.col("section").cast(pl.Int64, strict=False))
```

```python
acts_sections = acts_sections.join(cases_labeled.select("ddl_case_id"), on="ddl_case_id", how="semi")

acts_sections = acts_sections.join(act_key, on="act", how="left")

acts_sections = acts_sections.join(section_key, on="section", how="left")
```

```python
acts_sections.sink_parquet("../ddl_judicial_data/ddl_judicial_data_2018_processed/acts_sections.parquet")
```

```python
acts_sections.group_by("ddl_case_id").agg(pl.col("bailable_ipc").n_unique()).sort("bailable_ipc", descending=True).collect()
```

```python
acts_sections.group_by("ddl_case_id").agg(pl.col("criminal").n_unique()).sort("criminal", descending=True).collect()
```

```python
acts_sections.group_by("ddl_case_id").agg(pl.col("number_sections_ipc").n_unique()).sort("number_sections_ipc", descending=True).collect()
```

```python
acts_by_case = acts_sections.group_by("ddl_case_id").agg(pl.col(["act_s", "section_s"]), pl.col(["number_sections_ipc", "criminal", "bailable_ipc"]).drop_nulls().first(), pl.len().alias("n_acts_or_sections"))
```

```python
cases_labeled = cases_labeled.join(acts_by_case, on="ddl_case_id", how="left")

cases_labeled.select(pl.col(["act_s", "section_s", "number_sections_ipc", "criminal", "bailable_ipc", "n_acts_or_sections"]).null_count()).collect()
```
