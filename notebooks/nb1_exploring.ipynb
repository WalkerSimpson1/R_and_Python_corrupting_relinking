```python
judges_merge = pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/judge_case_merge_key.csv")
judges_clean = pl.scan_csv("../ddl_judicial_data/ddl_judicial_data_2018extracted/judges_clean.csv")

judges_clean = judges_clean.with_columns(pl.col(["start_date", "end_date"]).str.to_date(format="%Y-%m-%d", strict=False))

filing_judges_clean = judges_clean.select(pl.col(['ddl_judge_id', 'female_judge', 'start_date', 'end_date']).name.suffix("_filing"))
decision_judges_clean = judges_clean.select(pl.col(['ddl_judge_id', 'female_judge', 'start_date', 'end_date']).name.suffix("_decision"))
```

```python
cases_labeled = cases_labeled.join(judges_merge, on="ddl_case_id", how="left")

cases_labeled = cases_labeled.join(filing_judges_clean, left_on="ddl_filing_judge_id", right_on="ddl_judge_id_filing", how="left")
cases_labeled = cases_labeled.join(decision_judges_clean, left_on="ddl_decision_judge_id", right_on="ddl_judge_id_decision", how="left")
```
